### 获取多天的指标数据

`````python
# 初始化空的DataFrame用于存储结果
df = pd.DataFrame()

# 从今天倒推5天，获取日期列表
today = pd.Timestamp.now()
for i in range(5):
      # 计算日期
      target_date = today - pd.Timedelta(days=i)
      # 格式化为YYYYMMDD格式
      trade_date = target_date.strftime('%Y%m%d')
      print(trade_date)

      try:
            # 获取指定日期的数据
            daily_data = pro.daily_basic(ts_code="000001.SZ", trade_date=trade_date,
                                    fields='ts_code,trade_date,turnover_rate,volume_ratio,pe,pb')
            # 避免空数据或全NA列的concat警告
            if not daily_data.empty and not all(daily_data.isna().all()):
                  df = pd.concat([df, daily_data], ignore_index=True)
      except Exception as e:
      # 如果某一天没有数据，继续获取下一天
            continue
`````

`````python
def doubao_chat(prompt):
    import os
    from openai import OpenAI

    # 请确保您已将 API Key 存储在环境变量 ARK_API_KEY 中
    # 初始化Openai客户端，从环境变量中读取您的API Key
    client = OpenAI(
        # 此为默认路径，您可根据业务所在地域进行配置
        base_url="https://ark.cn-beijing.volces.com/api/v3",
        # 从环境变量中获取您的 API Key
        api_key=Doubao_API_KEY, # 这里你可以直接拷贝粘贴你自己的API Key
    )

    # Non-streaming:
    completion = client.chat.completions.create(
        # 指定您创建的方舟推理接入点 ID，此处已帮您修改为您的推理接入点 ID
        model="doubao-1-5-pro-32k-250115",
        messages=[
            {"role": "system", "content": "你是人工智能助手"},
            {"role": "user", "content": prompt},
        ],
    )
    return completion.choices[0].message.content
`````

### 从返回的response中提取json数据

`````python
def extract_json(response):
    """
    使用正则表达式，从返回的response中提取json数据
    """
    import re
    import json

    match = re.search(r'\{.*\}', response, re.DOTALL)
    if not match:
        raise ValueError("未在响应中找到 JSON 数据")
    json_str = match.group()
    return json.loads(json_str)

`````
