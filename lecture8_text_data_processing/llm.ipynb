### N-gram Models

- An N-gram is a contiguous sequence of 'N' items (or words) in a given text or sentence. 
- In the context of language modeling and natural language processing (NLP), N-grams are used to model the probability of a word (or item) occurring based on the history of the previous 'N-1' words.

Unigram (1-gram): Example: In the sentence "The cat is black," each word ("The," "cat," "is," "black") represents a unigram.

Bigram (2-gram): Example: In the sentence "The cat is black," possible bigrams include ("The cat," "cat is," "is black").

Trigram (3-gram):Example: In the sentence "The cat is black," possible trigrams include ("The cat is," "cat is black").

例如：“使机器能够理解语言一直是学术界的追求目标。然而，由于人类语言的复杂性和多变性，使机器能够进行语言翻译、语音识别、文字识别（无论是印刷体还是手写体）以及海量文献的自动检索，长久以来似乎是一个遥不可及的愿景。”

<font color = "red">数据稀疏性问题是一个常见的挑战：这意味着在训练数据中，
某些单词或短语可能从未出现过，或者其上下文信息不足，导致模型在估计这
些单词序列的概率时容易出现零概率问题。为了解决这一问题，引入了平滑
(smoothing)技术，它可以为那些在训练数据中未出现或出现次数极少的单词
或短语提供一个非零的概率估计，从而使模型的预测更为合理。</font>
