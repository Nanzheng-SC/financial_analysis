```python
# 首先确保您已经设置了API密钥
# 如果您使用.env文件，可以使用以下代码读取：
# from dotenv import load_dotenv
# import os
# load_dotenv()
# Doubao_API_KEY = os.getenv('Doubao_API_KEY')

# 如果没有设置API密钥，可以使用以下代码提供一个默认值（仅用于演示）
if 'Doubao_API_KEY' not in globals() or Doubao_API_KEY is None:
    print("警告：API密钥未设置，使用默认值进行演示")
    Doubao_API_KEY = "demo-api-key"  # 实际使用时需要替换为真实API密钥

# 定义增强版的doubao_chat函数，增加错误处理
def doubao_chat(prompt):
    import os
    from openai import OpenAI

    # 初始化Openai客户端
    client = OpenAI(
        base_url="https://ark.cn-beijing.volces.com/api/v3",
        api_key=Doubao_API_KEY,
    )

    try:
        completion = client.chat.completions.create(
            model="doubao-1-5-pro-32k-250115",
            messages=[
                {"role": "system", "content": "你是一个专业的文本处理助手，擅长关键词提取、摘要生成和中文分词。"},
                {"role": "user", "content": prompt},
            ],
        )
        return completion.choices[0].message.content
    except Exception as e:
        return f"API调用出错: {str(e)} （注意：实际运行时需要设置有效的API密钥）"

# 任务1: 提取新闻关键词
prompt1 = f"""请从以下新闻内容中提取10个最相关的关键词，以逗号分隔：
{news}
"""

# 任务2: 生成一句话新闻摘要
prompt2 = f"""请用一句话概括以下新闻的核心内容，不超过50个字：
{news}
"""

# 任务3: 对新闻进行切词
prompt3 = f"""请对以下新闻内容进行中文切词，并使用空格分隔各个词语：
{news}
"""

# 调用豆包API执行任务
print("=== 任务1: 提取关键词 ===")
keywords = doubao_chat(prompt1)
print(keywords)

print("\n=== 任务2: 一句话新闻摘要 ===")
summary = doubao_chat(prompt2)
print(summary)

print("\n=== 任务3: 新闻切词 ===")
word_segmentation = doubao_chat(prompt3)
print(word_segmentation[:500] + "..." if len(word_segmentation) > 500 else word_segmentation)
```

**如何管理会话记忆**

```python
# Initialize an empty list to store the conversation history
conversation_history = []

while True:
    # Get user input
    user_input = input("User: ")
    # Check if user wants to exit
    if user_input.lower() == 'exit':
        print("Chatbot: Goodbye!")
        break
    
    # Add the user's input to the conversation history
    conversation_history.append({"role": "user", "content": user_input})

    # Create a client instance
    client = OpenAI(
        # 此为默认路径，您可根据业务所在地域进行配置
        base_url="https://ark.cn-beijing.volces.com/api/v3",
        # 从环境变量中获取您的 API Key
        api_key=Doubao_API_KEY,
    )    
    # Call the chat.completions.create method with the conversation history
    response = client.chat.completions.create(
        model="doubao-1-5-pro-32k-250115",
        messages=conversation_history,
        temperature=0.95,
    )
    
    # Output ChatGLM’s response
    s = response.choices[0].message.content
    print("Chatbot:", s)
    
    # Add the chatbot's response to the conversation history
    conversation_history.append({"role": "assistant", "content": s})
```
