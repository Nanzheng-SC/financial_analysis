## 矩阵相乘（包括标量与向量相乘）

```python
import numpy as np
```

```python
# 创建一个矩阵
maxtrix = np.array(
      [[0.3, 0.4, 0.2, 0.7, 0.2],
       [0.5, 0.4, 0.7, 0.7, 0.1],
       [0.7, 0.6, 0.2, 0.7, 0.2]
])
# 创建一个向量
vec1 = np.array([0.2, 0.5, 0.1, 0.8, 0.3])
```

```python
# 计算每个行向量长度（2范数）
matrix_norm = np.linalg.norm(maxtrix, axis=1)
print(matrix_norm.shape)
# 计算vec1向量的长度
vec1_norm = np.linalg.norm(vec1)
print(vec1_norm.shape)
# 下面dot可以使用*号，也可以使用np.dot()函数
np.dot(matrix_norm, vec1_norm)
```

```python
# 创建一个二维数组
arr2 = np.array(
      [[1, 2], 
      [3, 4], 
      [5, 6]])

# 标量乘法
result2 = 0.5 * arr2
print("二维数组标量乘法:")
print(result2)
```

```python
# 计算每个文档与向量的点积
matrix_vec_doct = np.dot(maxtrix, vec1)
# 计算每个文档与向量的余弦相似度
# 分母：每个文档向量长度 * 向量长度，分子：每个文档与向量的点积
cosine_sim = matrix_vec_doct / np.dot(matrix_norm, vec1_norm)
print(cosine_sim)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cosine_similarity(maxtrix, vec1.reshape(1, -1))
```
