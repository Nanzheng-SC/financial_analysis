## 矩阵相乘（包括标量与向量相乘）
- 上节课没有特别关注一个标量和一个向量相乘

标量与向量的乘法可以用以下LaTeX格式表示：

### 定义与运算规则
设标量为 $k$，向量为 $\mathbf{v} = (v_1, v_2, \ldots, v_n)$，则标量乘以向量的结果为：

$k \cdot \mathbf{v} = (k \cdot v_1, k \cdot v_2, \ldots, k \cdot v_n)$

### 数学表达式
更严格地，我们可以写成：

$k \cdot \begin{pmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{pmatrix} = \begin{pmatrix} k \cdot v_1 \\ k \cdot v_2 \\ \vdots \\ k \cdot v_n \end{pmatrix}$

### 几何表示
从几何角度，这一运算相当于：

$k \cdot \vec{v} = |k| \cdot \vec{v} \cdot \text{sgn}(k)$

其中 $\text{sgn}(k)$ 表示 $k$ 的符号（决定方向是否反转）。

### 代数性质
标量乘法满足以下性质：

1. **分配律**：
   - $k(\mathbf{u} + \mathbf{v}) = k\mathbf{u} + k\mathbf{v}$
   - $(k + m)\mathbf{v} = k\mathbf{v} + m\mathbf{v}$

2. **结合律**：
   - $k(m\mathbf{v}) = (km)\mathbf{v}$

3. **单位元**：
   - $1 \cdot \mathbf{v} = \mathbf{v}$

### 示例
当 $k = 3$ 且 $\mathbf{v} = (2, -1, 4)$ 时：

$3 \cdot \begin{pmatrix} 2 \\ -1 \\ 4 \end{pmatrix} = \begin{pmatrix} 3 \cdot 2 \\ 3 \cdot (-1) \\ 3 \cdot 4 \end{pmatrix} = \begin{pmatrix} 6 \\ -3 \\ 12 \end{pmatrix}$
