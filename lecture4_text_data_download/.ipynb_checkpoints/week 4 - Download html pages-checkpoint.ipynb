## Escape character

[转义字符](https://ai4science.feishu.cn/wiki/Bnpow6FsziP9v5kxWz7cwmQNnEc)

python的转义字符是\，可以转义很多字符，比如\n表示换行，\t表示制表符，字符\本身也要转义，所以\\表示的字符就是\。Python还允许用“r”表示内部的字符串默认不转义。

```python
path = "c:\towhere"
print(path)
```

```python
path = r"c:\towhere"
print(path)
```

```python
path = "c:\nowhere.py"
print (path)
```

```python
p = r"c:\nowhere.py"
print(p)
```

除了使用r，还可以直接使用\,以便在字符串中包含原本的转义字符

```python
p = 'c:\\nowhere.py'
print(p)
```

## 编码与解码：encode and decode

### ASCII and unicode

**ASCII** stands for American Standard Code for Information Interchange. Computers can only understand numbers, so an ASCII code is the numerical representation of a character such as 'a' or '0' or an action of some sort.

<img src="../img/ascii.png" align='left' width = 1500 height = 800/>

```python
0x61
```

```python
"\x61"
```

```python
chr(0x61)
```

```python
chr(0x63)
```

```python
ord("中")
```

```python
hex(97) # 十进制-》十六进制
```

```python
chr(97)
```

```python
# ASCII 编码 (只能处理基本英文字符)
ascii_text = "Hello, World!"
encoded_ascii = ascii_text.encode('ascii')
print(f"ASCII 编码: {encoded_ascii}")
```

```python
byte_string = b'Hello, World!'
# 查看每个字节的十进制ASCII值
print(list(encoded_ascii))  
```

```python
# 查看每个字节的十六进制表示
print([hex(b) for b in byte_string])  # 输出十六进制列表
```

0x48 = 4×16¹ + 8×16⁰ = 4×16 + 8×1 = 64 + 8 = 72

```python
# 解码
decoded_ascii = encoded_ascii.decode('ascii')
print(decoded_ascii)
```

大家观察分析下面的代码和输出

```python
text_cn = "中文"
try:
    text_cn.encode("ascii")  # ASCII 无法编码中文
except UnicodeEncodeError as e:
    print(f"编码失败: {e}")
```

### encoding和decoding总结

Unicode把所有语言都统一到一套编码里(2字节），这样就不会再有乱码问题了。但是，用Unicode编码比ASCII编码需要更多的存储空间，在存储和传输上就十分不划算。又出现了把Unicode编码转化为“可变长编码”的UTF-8编码。更多资料可以参考[字符编码](https://www.liaoxuefeng.com/wiki/1016959663602400/1017075323632896)。在计算机内存中，统一使用Unicode编码，当需要保存到硬盘或者需要传输的时候，就转换为其他编码：
* 将Unicode字符串编码为字节(encode)
* 将字节解码为Unicode字符串(decode)



```python
'中国'.encode('utf-8')
```

```python
# encoding一个字符串
text = "数据分析"
b = text.encode("gbk")
b
```

```python
# encodicng一个字符串
c = text.encode("utf-8")
c
```

```python
b.decode("utf-8")
```

### chardet

字符串编码一直是令人非常头疼的问题，尤其是我们在处理一些不规范的第三方网页的时候。虽然Python提供了Unicode表示的str和bytes两种数据类型，并且可以通过encode()和decode()方法转换，但是，在不知道编码的情况下，对bytes做decode()不好做。

对于未知编码的bytes，要把它转换成str，需要先“猜测”编码。猜测的方式是先收集各种编码的特征字符，根据特征字符判断，就能有很大概率“猜对”。

当然，我们肯定不能从头自己写这个检测编码的功能，这样做费时费力。chardet这个第三方库正好就派上了用场。用它来检测编码，简单易用。

如果安装了Anaconda，chardet就已经可用了。否则，需要在命令行下通过pip安装：

pip install chardet

```python
import chardet
```

```python
data = '数据分析'.encode('gbk')
```

```python
data
```

```python
chardet.detect(data)
```

```python
with open("example.txt","rb") as f:
    r = f.read()
r
```

```python
chardet.detect(r)
```

```python
r.decode("utf-8")
```

# 如何处理服务器拒绝请求的问题

### 添加请求头，模拟Chrome浏览器

```python
import requests

url = "http://guba.eastmoney.com/list,zssh000001.html"

# 添加请求头，模拟Chrome浏览器
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# 使用带请求头的get请求
response = requests.get(url, headers=headers)

# 检查响应状态码
print(response.status_code)  # 应该返回200表示成功
```

We can read the content of the server’s response.

Requests will automatically decode content from the server. Most unicode charsets are seamlessly decoded.

```python
# We can read the content of the server’s response
html =  response.text
print(html)
```

```python
response.content
```

response.text 与response.content的区别
response.text返回的是处理过的Unicode型的数据，而使用response.content返回的是bytes型的原始数据。也就是说，response.content相对于response.text来说节省了计算资源，response.content是把内容bytes返回.

而response.text是decode成Unicode.


When you make a request, Requests makes educated guesses about the encoding of the response based on the **HTTP headers**. The text encoding guessed by Requests is used when you access r.text. You can find out what encoding Requests is using, and change it, using the r.encoding property:

If you change the encoding, Requests will use the new value of r.encoding whenever you call r.text.

```python
response.encoding
```

```python
# r.apparent_encoding 从内容中分析出的响应内容编码方式
response.apparent_encoding
```

- 通过 response.encoding = response.apparent_encoding ，我们强制requests库使用基于内容分析猜测的编码来解码响应内容
- 这可以有效解决因服务器响应头中编码声明不正确而导致的乱码问题

```python
response.encoding = response.apparent_encoding
```

## 写个函数，实现下载一个页面

```python
def download_page(url,para = None):
    import requests
    headers = {'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/52.0.2743.116 Safari/537.36 Edge/15.15063'}
    if para:
        response = requests.get(url,params=para,headers = headers)
    else:
        response = requests.get(url,headers = headers)
    response.encoding = response.apparent_encoding
    if response.status_code == 200:
        return response.text
    else:
        return "failed"
```

```python
url = "https://www.imdb.com/title/tt0110912/reviews"
pa = {"ratingFilter":"1"}
download_page(url,pa)
```

## 课后作业
上证指数吧第一页的url = "http://guba.eastmoney.com/list,zssh000001_1.html"

tips - you can contruct a url for each page

```python
import time
import requests
from bs4 import BeautifulSoup
import random
import json
import re

def download_page(url, max_retries=3):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    for attempt in range(max_retries):
        try:
            response = requests.get(url, headers=headers, timeout=10)
            if response.status_code == 200:
                return response.text
            else:
                print(f"请求失败，状态码: {response.status_code}")
        except requests.exceptions.RequestException as e:
            print(f"请求异常 (尝试 {attempt + 1}/{max_retries}): {e}")
            if attempt < max_retries - 1:
                time.sleep(random.uniform(2, 5))  # 等待时间稍长一些
                continue
        return ""

def parse_eastmoney_guba(html_content):
    # 使用正则表达式提取 article_list 变量中的JSON数据
    pattern = r'var article_list=(\{.*?\});'
    match = re.search(pattern, html_content, re.DOTALL)
    
    if match:
        json_str = match.group(1)
        try:
            data = json.loads(json_str)
            return data
        except json.JSONDecodeError as e:
            print(f"JSON解析错误: {e}")
            return None
    else:
        print("未找到article_list数据")
        return None

# 创建URL列表
url_list = [f"http://guba.eastmoney.com/list,zssh000001_{i}.html" for i in range(1, 10)]

# 创建空列表存储所有数据
all_data = []

for i, u in enumerate(url_list, 1):
    print(f"正在下载第 {i} 页: {u}")
    r = download_page(u)
        
    data = parse_eastmoney_guba(r)
    
    if data:
        print(f"第 {i} 页成功解析，获取到 {len(data.get('re', []))} 条帖子")
        all_data.append(data)
    else:
        print(f"第 {i} 页解析失败")
    
    # 随机 sleep 1-3 秒
    sleep_time = random.uniform(0.5, 1)
    print(f"等待 {sleep_time:.2f} 秒后继续...")
    time.sleep(sleep_time)

# 打印汇总信息
print(f"\n下载完成！成功获取 {len(all_data)} 页数据")
for i, data in enumerate(all_data, 1):
    if data and 're' in data:
        print(f"第 {i} 页有 {len(data['re'])} 条帖子")
```

```python
import csv

def extract_comment_data(data_list):
    """
    提取多个字典中包含的评论数据，每个字典都有're'键
    
    参数:
        data_list: 包含多个字典的列表，每个字典都有're'键
                   而're'键的值是包含评论数据的列表
    
    返回:
        提取后的评论信息列表
    """
    extracted_data = []
    
    # 遍历列表中的每个字典
    for item in data_list:
        # 从每个字典的're'键中获取评论列表
        comments = item.get('re', [])
        
        # 遍历当前字典中的每条评论
        for comment in comments:
            # 提取关键字段
            comment_info = {
                '帖子ID': comment.get('post_id'),
                '帖子标题': comment.get('post_title'),
                '股票吧代码': comment.get('stockbar_code'),
                '股票吧名称': comment.get('stockbar_name'),
                '用户ID': comment.get('user_id'),
                '用户昵称': comment.get('user_nickname'),
                '点击量': comment.get('post_click_count'),
                '转发量': comment.get('post_forward_count'),
                '评论量': comment.get('post_comment_count'),
                '发布时间': comment.get('post_publish_time'),
                '最后更新时间': comment.get('post_last_time'),
                '是否有图片': comment.get('post_has_pic'),
                '是否有视频': comment.get('post_has_video')
            }
            
            extracted_data.append(comment_info)
    
    return extracted_data

def save_to_csv(data, filename='comments.csv'):
    """
    将提取的评论数据保存到CSV文件
    
    参数:
        data: 提取后的评论信息列表
        filename: 保存的CSV文件名，默认为'comments.csv'
    """
    if not data:
        print("没有数据可保存到CSV文件")
        return
    
    # 获取表头（字典的键）
    fieldnames = data[0].keys()
    
    try:
        # 使用utf-8-sig编码确保中文正常显示
        with open(filename, 'w', newline='', encoding='utf-8-sig') as csvfile:
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            
            # 写入表头
            writer.writeheader()
            
            # 写入数据行
            for row in data:
                writer.writerow(row)
        
        print(f"成功将 {len(data)} 条评论数据保存到 {filename}")
    except Exception as e:
        print(f"保存CSV文件时出错: {str(e)}")

# 示例数据 - 包含多个字典，每个字典都有're'键
sample_data = all_data

# 提取数据
extracted_comments = extract_comment_data(sample_data)

# 打印结果
print(f"共提取到 {len(extracted_comments)} 条评论数据：")
print("-" * 80)
# for i, comment in enumerate(extracted_comments, 1):
#     print(f"评论 {i}:")
#     for key, value in comment.items():
#         print(f"  {key}: {value}")
#     print("-" * 80)

# 保存到CSV文件
save_to_csv(extracted_comments)
    
```
