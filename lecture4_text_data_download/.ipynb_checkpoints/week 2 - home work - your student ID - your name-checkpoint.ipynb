# develop a Naive Classifier  (100 points)

- Data: comments, Dictionary, Lexicon
- Each word in the lexicon is identified as being positive, negative 
- Each word in a message is checked against the lexicon, and assigned a value (-1, +1) based on the default value in the lexicon
- the net word count of all lexicon matched words is taken, and if this value is greater than one, we sign the message as a buy, return 1. If this value is less than negative one, the message is taken to be a sell, return -1, else return 0. 
- you can use the function cut to tokenize the comments text. 

```python
url = "http://guba.eastmoney.com/list,zssh000001.html"
```

```python
import re
def download_page(url,para = None):
    import requests
    headers = {'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/52.0.2743.116 Safari/537.36 Edge/15.15063'}
    if para:
        response = requests.get(url,params=para,headers = headers)
    else:
        response = requests.get(url,headers = headers)
    response.encoding = response.apparent_encoding
    if response.status_code == 200:
        return response.text
    else:
        print ("failed to download the page")
```

```python
# get all the conmments in the first page
html = download_page(url)
pattern = r"""(<div class="articleh normal_post.*?">.*?</div>)"""
comments_list = re.findall(pattern,html,re.DOTALL)
title_pattern = r"""<span class="l3 a3">.*?<a .*?href=".*?" title="(.*?)">.*?</a></span>"""
comments = []
for c in comments_list:
    t = re.findall(title_pattern,c)  
    comments.append(t[0])
```
