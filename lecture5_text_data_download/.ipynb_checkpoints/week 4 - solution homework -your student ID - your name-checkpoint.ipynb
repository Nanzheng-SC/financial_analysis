# 数据抓取与可视化（50 points）
- 使用Beautifulsoup抓取上证指数吧前20页的数据(只下载标题和时间）；
- 对帖子数量每个小时的时间分布做简单可视化。
- 下面部分代码供参考

```python
import re
from bs4 import BeautifulSoup
import requests
```

```python
links = ["http://guba.eastmoney.com/list,zssh000001_{}.html".format(i) for i in range(1,21)]
```

```python
links
```

```python
def download_page(url,para = None):
    headers = {'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/52.0.2743.116 Safari/537.36 Edge/15.15063'}
    if para:
        response = requests.get(url,params=para,headers = headers)
    else:
        response = requests.get(url,headers = headers)
    response.encoding = response.apparent_encoding
    if response.status_code == 200:
        return response.text
    else:
        return None
```

```python
comments = []
titles = []
dates = []
for link in links:
    html = download_page(link)
    soup = BeautifulSoup(html)

    class_pattern = re.compile("articleh normal_post.*?")
    all_comments = soup.find_all("div",{'class':class_pattern})
    

    for c in all_comments:
        titles.append(c.find("span",{"class":"l3 a3"}).text)
        dates.append(c.find("span",{"class":"l5 a5"}).text)
    
for index,c in enumerate(zip(titles,dates)):
    comments.append((c[0],c[1]))
```

```python
len(comments)
```

```python
comments[:4]
```

```python
def get_hour(date):
    from dateutil import parser
    date = parser.parse("{}".format(date))
    return date.hour
```

```python
get_hour(comments[2][1])
```

```python
newsFreq = {}
newsFreq[0] = 2
newsFreq[3] = 20
newsFreq.get(4,0)
```

```python
#统计每天的新闻数量
newsFreq = {}
for c in comments:
    hour = get_hour(c[1])
    newsFreq[hour] = newsFreq.get(hour, 0) + 1
print(newsFreq)
```

```python
newsList = list(newsFreq.items())
newsList.sort(key=lambda x: x[0])
newsList
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure(figsize=(5, 5))
dateList = [item[0] for item in newsList]
freqList = [item[1] for item in newsList]
#显示柱状图
plt.bar(range(len(dateList)), freqList, tick_label=dateList)
```
