```python
# html_text是一串html文档，类似上面download所得到的网页
html = """
<html>
 <head>
<title>腾讯首页</title>
 </head>
 <body>
    <li class="nav-item">
    <a href="http://news.qq.com/" target="_blank" bosszone="dh_1">新闻</a>
  </li>
    <li class="nav-item">
    <a href="http://v.qq.com/" target="_blank" bosszone="dh_2">视频</a>
  </li>
 </body>
</html> 
"""
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(html)

for a in soup.find_all("a"):
    print (a)
```

```python
# 如何使用BeautifulSoup解析出上面html中的超链接呢？
for a in soup.find_all("a"):
    print (a.get("href"))
```
