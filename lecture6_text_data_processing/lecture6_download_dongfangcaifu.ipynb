# 数据下载并存储

```python
import requests
import json
```

```python
import re
def download_page(url,para = None):
    headers = {'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/52.0.2743.116 Safari/537.36 Edge/15.15063'}
    if para:
        response = requests.get(url,params=para,headers = headers)
    else:
        response = requests.get(url,headers = headers)
    response.encoding = response.apparent_encoding
    if response.status_code == 200:
        return response.text
    else:
        print ("failed to download the page")
```

```python
def parse_eastmoney_guba(html_content):
    """
    从 EastMoney 股吧 HTML 内容中提取 article_list(股吧评论) 变量中的 JSON 数据。
    
    参数:
        html_content (str): EastMoney 股吧页面的 HTML 内容。
    
    返回:
        dict or None: 解析后的 JSON 数据字典，如果解析失败则返回 None。
    """
    pattern = r'var article_list=(\{.*?\});'
    match = re.search(pattern, html_content, re.DOTALL)
    
    if match:
        json_str = match.group(1)
        try:
            data = json.loads(json_str)
            return data
        except json.JSONDecodeError as e:
            print(f"JSON解析错误: {e}")
            return None
    else:
        print("未找到article_list数据")
        return None
```

```python
url_list = ["https://guba.eastmoney.com/list,zssh000001.html",
           "https://guba.eastmoney.com/list,zssh000001_2.html",
           "https://guba.eastmoney.com/list,zssh000001_3.html",
           "https://guba.eastmoney.com/list,zssh000001_4.html",
           "https://guba.eastmoney.com/list,zssh000001_5.html"]
```

```python
# get all the conmments in the first page，大家要下载前3页的评论数据
# 这里仅仅用于测试使用，大家不需要可以删除
url = url_list[0]
html = download_page(url)
comments = parse_eastmoney_guba(html)
for comment in comments["re"]:
    print(comment)
```

```python
comments_list = []
for url in url_list:
    html = download_page(url)
    comments = parse_eastmoney_guba(html)
    for c in comments.get("re"):
        comments_list.append(c.get("post_last_time") + " " + c.get("post_title") + "\n")
```

```python
comments_list
```

```python
# 打开文件（若不存在则创建，'w' 表示覆盖写入，'a' 表示追加写入）
with open("sse_comments.txt", "w", encoding="utf-8") as f:
    for text in comments_list:
        # 每个元素后加换行符 '\n'，确保换行
        f.write(text)
```
